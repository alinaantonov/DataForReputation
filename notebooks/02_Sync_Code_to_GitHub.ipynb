# Publish research code to GitHub (not datasets)

**Safe-default workflow:** Sync only the framework's source code/notebooks/docs/tests. Never publish `runs/` or original notebooks containing keys. The publisher performs a basic credential scan and checks for remote file conflicts.

**Prerequisites:** Create an existing GitHub repository; add a Colab Secret named `GITHUB_TOKEN` (minimum suitable repository permissions); enable notebook access to the secret. Rotate/revoke any previously exposed personal access token. The full repository history may require separate secret-removal work; this publisher does not clean Git history.

Run a **dry run first**. Set `DRY_RUN=False` only after inspecting file changes. If this repository already has code on the very first sync, `ALLOW_INITIAL_OVERWRITE=True` permits replacing matching paths; review the preview first. Avoid editing the same files independently in GitHub and Drive.

```python
from google.colab import drive, userdata
from pathlib import Path
import sys, json

drive.mount('/content/drive')
PROJECT_ROOT = Path('/content/drive/MyDrive/ColabNotebooks/DataForReputation')
sys.path.insert(0, str(PROJECT_ROOT))
from tools.sync_code_to_github import audit_code_files, publish_code
```

```python
# Review files selected for upload, without using any credentials.
files = audit_code_files(PROJECT_ROOT)
for file in files:
    print(file.relative_to(PROJECT_ROOT))
```

```python
'''
GITHUB_USERNAME = "alinaantonov"
GITHUB_REPOSITORY = "DataForReputation"
BRANCH = "main"
GIT_USER_EMAIL = "alina_antonov@yahoo.com"
https://github.com/alinaantonov/DataForReputation.git

userdata.get('secretName')
'''

# Configure repository and mode.
YOUR_USERNAME = 'alinaantonov'
GITHUB_REPO = YOUR_USERNAME + '/DataForReputation'
DRY_RUN = False  # default: preview only
ALLOW_INITIAL_OVERWRITE = True  # change ONLY if first-run preview warrants it
COMMIT_MESSAGE = 'Update reputation extraction framework'

if 'YOUR_USERNAME' in GITHUB_REPO:
    raise ValueError('Enter your real GitHub username/repository first.')
TOKEN = userdata.get('GITHUB_TOKEN')
result = publish_code(
    root=PROJECT_ROOT,
    repo_slug=GITHUB_REPO,
    token=TOKEN,
    message=COMMIT_MESSAGE,
    dry_run=DRY_RUN,
    allow_initial_overwrite=ALLOW_INITIAL_OVERWRITE,
)
print(result['message'])
print('Changed files:', result['changed'], '| conflicts:', result['conflicts'])
for item in result['files']:
    if item['status'] != 'unchanged':
        print(item['status'], item['file'])
if result.get('commit'):
    print('Commit:', result['commit'])
```
